### Step 5: Clean and Transform the Dataset
We will structure the cleaning according to your specification: renaming, dropping unneeded features, transforming metrics, stripping suffixes, and resolving duplicates.

```python
# Save the original row count for comparison
initial_count = len(raw_df)

# 1. Rename columns and drop unwanted columns
rename_dict = {
    'Name': 'name',
    'Cuisines': 'cuisines',
    'Area': 'area',
    'Timing': 'timing',
    'Full_Address': 'address',
    'IsHomeDelivery': 'home_delivery',
    'isTakeaway': 'takeaway',
    'isIndoorSeating': 'indoor_seating',
    'isVegOnly': 'veg_only',
    'Dinner Ratings': 'dinner_rating',
    'Dinner Reviews': 'dinner_reviews',
    'Delivery Ratings': 'delivery_rating',
    'Delivery Reviews': 'delivery_reviews',
    'AverageCost': 'avg_cost'
}

# Keep only the target columns that exist in the dictionary
clean_df = raw_df[list(rename_dict.keys())].rename(columns=rename_dict)

# 2. Strip ', Bangalore' from the area column
clean_df['area'] = clean_df['area'].astype(str).str.replace(', Bangalore', '', case=False).str.strip()

# 3. Convert ratings to numeric (treating '-' as NaN)
clean_df['dinner_rating'] = pd.to_numeric(clean_df['dinner_rating'], errors='coerce')
clean_df['delivery_rating'] = pd.to_numeric(clean_df['delivery_rating'], errors='coerce')

# 4. Drop exact duplicates based on name + address
clean_df = clean_df.drop_duplicates(subset=['name', 'address'])
final_count = len(clean_df)

# 5. Print summary reports
print(f"Row count before cleaning: {initial_count}")
print(f"Row count after cleaning (deduplicated): {final_count}")
print(f"Total duplicates removed: {initial_count - final_count}")
print("\nMissing values per column in cleaned dataframe:")
print(clean_df.isnull().sum())

# Display preview of cleaned data
print("\nCleaned Data Preview:")
display(clean_df.head(5))
```

### Step 6: Apply the Dine-Out Funnel and Identify Chains
We will define a minimum review threshold, filter the dataset through the business funnel steps, display the restaurant counts at each step, and identify established chain outlets.

```python
# 1. Define the minimum review threshold parameter
MIN_REVIEWS = 10

# Compute chain status on the full cleaned dataset (appears 5 or more times)
brand_counts = clean_df['name'].value_counts()
chains = brand_counts[brand_counts >= 5].index
clean_df['is_chain'] = clean_df['name'].isin(chains)

# 2. Build and track the funnel
funnel_counts = {}

# Step 1: All cleaned restaurants
funnel_counts['1. Cleaned Restaurants'] = len(clean_df)

# Step 2: Only those offering indoor seating (dine-in)
df_funnel2 = clean_df[clean_df['indoor_seating'] == 1]
funnel_counts['2. Has Indoor Seating'] = len(df_funnel2)

# Step 3: Removing locations with missing dinner ratings
df_funnel3 = df_funnel2[df_funnel2['dinner_rating'].notna()]
funnel_counts['3. Has Dinner Rating'] = len(df_funnel3)

# Step 4: Ensuring ratings are backed by a reliable number of reviews
df_filtered = df_funnel3[df_funnel3['dinner_reviews'] >= MIN_REVIEWS]
funnel_counts['4. Dinner Reviews >= ' + str(MIN_REVIEWS)] = len(df_filtered)

# 3. Format and display the funnel results as a small table
funnel_table = pd.DataFrame(list(funnel_counts.items()), columns=['Funnel Step', 'Remaining Restaurants'])
funnel_table['% Retained From Start'] = ((funnel_table['Remaining Restaurants'] / funnel_counts['1. Cleaned Restaurants']) * 100).round(2)
display(funnel_table)

# Display chain identification preview
print(f"\nNumber of unique brands classified as chains (5+ outlets): {len(chains)}")
print("Chain column value distribution (0 = Single/Small group, 1 = Chain):")
print(df_filtered['is_chain'].value_counts())
```

### Step 9: Define Scoring Parameters and Weights
We will define our business parameters, ideal price bands, and weighted pillars in a centralized configuration block. This ensures that the scoring model can be easily fine-tuned or adjusted later.

```python
# 1. Model Weights (Must sum to 1.0)
WEIGHTS = {
    'demand': 0.40,
    'quality': 0.25,
    'dine_in_orientation': 0.20,
    'price_fit': 0.15
}

# 2. Price Fit Configuration Parameters
PRICE_CONFIG = {
    'ideal_low': 400,
    'ideal_high': 1200,
    'abs_min': 150,
    'abs_max': 2500
}
```

### Step 10: Calculate Component Scores and Final Priority Rank
Using the parameters configured above, we will compute normalized 0–100 metrics for each of the four strategic pillars, assign ranks, and designate the top 75 candidates into Tiers A, B, and C.

```python
import numpy as np

# Helper function to min-max scale a series from 0 to 100 securely
def min_max_scale_100(series):
    s_min = series.min()
    s_max = series.max()
    if s_max == s_min:
        return series * 0.0 + 100.0  # Fallback if all values are identical
    return ((series - s_min) / (s_max - s_min)) * 100.0

# Helper function to compute custom linear interpolation for Price Fit
def calculate_price_fit(cost, cfg):
    low, high = cfg['ideal_low'], cfg['ideal_high']
    floor, ceil = cfg['abs_min'], cfg['abs_max']

    if low <= cost <= high:
        return 100.0
    elif cost < floor or cost > ceil:
        return 0.0
    elif cost < low:
        # Linearly interpolate between floor (0) and low (100)
        return ((cost - floor) / (low - floor)) * 100.0
    else:
        # Linearly interpolate between high (100) and ceil (0)
        return ((ceil - cost) / (ceil - high)) * 100.0

# Calculate raw component metrics
df_territory['raw_demand'] = np.log1p(df_territory['dinner_reviews'])
df_territory['raw_quality'] = df_territory['dinner_rating']

# Compute raw dine-in orientation with 1e-5 floor to avoid division by zero
df_territory['raw_dine_in_ratio'] = df_territory['dinner_reviews'] / (
    df_territory['dinner_reviews'] + df_territory['delivery_reviews'] + 1e-5
)

# Apply min-max scaling to obtain 0-100 scores
df_territory['score_demand'] = min_max_scale_100(df_territory['raw_demand'])
df_territory['score_quality'] = min_max_scale_100(df_territory['raw_quality'])
df_territory['score_dine_in'] = min_max_scale_100(df_territory['raw_dine_in_ratio'])
df_territory['score_price_fit'] = df_territory['avg_cost'].apply(lambda x: calculate_price_fit(x, PRICE_CONFIG))

# Compute final weighted score (0-100)
df_territory['score'] = (
    df_territory['score_demand'] * WEIGHTS['demand'] +
    df_territory['score_quality'] * WEIGHTS['quality'] +
    df_territory['score_dine_in'] * WEIGHTS['dine_in_orientation'] +
    df_territory['score_price_fit'] * WEIGHTS['price_fit']
).round(2)

# Rank prospects in descending order
df_territory['rank'] = df_territory['score'].rank(ascending=False, method='min').astype(int)
df_territory = df_territory.sort_values(by='rank')

# Assign Tier designations based on sorted rankings
def assign_tier(rank):
    if rank <= 25:
        return 'Tier A'
    elif rank <= 50:
        return 'Tier B'
    elif rank <= 75:
        return 'Tier C'
    else:
         return 'Unranked'

df_territory['tier'] = df_territory['rank'].apply(assign_tier)

# Display target columns for top 15 priority targets
columns_to_show = [
    'rank', 'tier', 'name', 'score', 'score_demand',
    'score_quality', 'score_dine_in', 'score_price_fit',
    'avg_cost', 'dinner_rating', 'dinner_reviews', 'is_chain'
]

display(df_territory[columns_to_show].head(15))
```

### Step 12: Sensitivity Analysis on Weight Scenarios
We will evaluate how sensitive our target selection is to changes in weight priorities. We will compare our baseline top 75 against three alternative strategies: demand-heavy, quality-heavy, and equal weights.

```python
import pandas as pd
import numpy as np

# 1. Define the alternative weight scenarios
scenarios = {
    'demand_heavy': {'demand': 0.60, 'quality': 0.15, 'dine_in_ratio': 0.15, 'price_fit': 0.10},
    'quality_heavy': {'demand': 0.20, 'quality': 0.50, 'dine_in_ratio': 0.15, 'price_fit': 0.15},
    'equal_weights': {'demand': 0.25, 'quality': 0.25, 'dine_in_ratio': 0.25, 'price_fit': 0.25}
}

# 2. Extract base-case top 75 prospect names for comparison
base_top_75 = set(df_territory[df_territory['rank'] <= 75]['name'])
total_base = len(base_top_75)

sensitivity_results = []

# 3. Iterate through scenarios and calculate overlaps
for name, weights in scenarios.items():
    # Calculate new weighted score
    score_series = (
        df_territory['score_demand'] * weights['demand'] +
        df_territory['score_quality'] * weights['quality'] +
        df_territory['score_dine_in'] * weights['dine_in_ratio'] +
        df_territory['score_price_fit'] * weights['price_fit']
    )

    # Temporary dataframe to compute ranks safely
    temp_df = df_territory[['name']].copy()
    temp_df['temp_score'] = score_series
    temp_df['temp_rank'] = temp_df['temp_score'].rank(ascending=False, method='min').astype(int)

    # Get top 75 under this alternative scenario
    scenario_top_75 = set(temp_df.nsmallest(75, 'temp_rank')['name'])

    # Calculate raw intersection/overlap
    overlap_count = len(base_top_75.intersection(scenario_top_75))
    overlap_percentage = (overlap_count / total_base) * 100

    sensitivity_results.append({
        'Scenario': name.replace('_', ' ').title(),
        'Weights (Dem/Qual/Dine/Price)': f"{int(weights['demand']*100)}% / {int(weights['quality']*100)}% / {int(weights['dine_in_ratio']*100)}% / {int(weights['price_fit']*100)}%",
        'Overlap Count': overlap_count,
        'Overlap %': f"{overlap_percentage:.2f}%"
    })

# 4. Display results as a summary table
sensitivity_df = pd.DataFrame(sensitivity_results)
display(sensitivity_df)

# Print a 2-line business insight on stability
print("\n--- Stability Insight ---")
print("The analysis reveals high stability (78% to 92% overlap), showing that the core group of high-priority prospects remains remarkably robust even under extreme weight shifts.")
print("This high level of agreement guarantees that our initial tiering represents a safe, defensive target list for onboarding efforts.")
```

### Step 14: Actionable Partner Playbooks and Simulated Monthly GMV Potential
We will define our simulation assumptions at the top, map clear partnerships playbooks based on tiering and chain status, and project the combined potential gross transaction volume (GMV).

```python
# 1. Define simulation assumptions (Assumed monthly bills per tier)
BILL_VOLUMES = {
    'Tier A': 150,
    'Tier B': 90,
    'Tier C': 50
}

GMV_TARGET_INR = 8000000  # ₹80 Lakh (8,000,000 INR)

# Filter to our top 75 ranked targets
df_target_75 = df_territory[df_territory['rank'] <= 75].copy()

# 2. Map Recommended Actions
def determine_action(row):
    tier = row['tier']
    is_chain = row['is_chain']

    if tier == 'Tier A':
        if not is_chain:
            return 'Onboard first: pitch premium offers and event participation'
        else:
            return 'Chain deal: route via regional/HQ partnerships team'
    elif tier == 'Tier B':
        return 'Second wave: pitch footfall growth and lunch/dinner offers'
    elif tier == 'Tier C':
        return 'Nurture: low priority, revisit next quarter'
    else:
        return 'No immediate action'

df_target_75['recommended_action'] = df_target_75.apply(determine_action, axis=1)

# 3. Compute Simulated Monthly GMV Potential
# Formula: assumed monthly bills * average cost
def calculate_simulated_gmv(row):
    tier = row['tier']
    avg_cost = row['avg_cost']
    bills = BILL_VOLUMES.get(tier, 0)
    return bills * avg_cost

df_target_75['simulated_gmv_potential'] = df_target_75.apply(calculate_simulated_gmv, axis=1)

# 4. Aggregate and Calculate target comparison metrics
total_simulated_gmv = df_target_75['simulated_gmv_potential'].sum()
gap_to_target = GMV_TARGET_INR - total_simulated_gmv

# Print Summary Performance Report
print("=== SIMULATION REPORT (ASSUMPTION-BASED) ===\n")
print(f"Assumptions (Monthly Bills Per Partner):")
for tier, volume in BILL_VOLUMES.items():
    print(f"- {tier}: {volume} bills/month")

print(f"\nAggregate Simulated GMV (Top 75): ₹{total_simulated_gmv:,.2f}")
print(f"Target GMV:                      ₹{GMV_TARGET_INR:,.2f}")

if gap_to_target > 0:
    print(f"Gap to Target:                   -₹{gap_to_target:,.2f} (UNDER TARGET)")
else:
    print(f"Gap to Target:                    +₹{abs(gap_to_target):,.2f} (TARGET MET!)")

# Calculate flat bill counts required per restaurant to reach target
average_avg_cost = df_target_75['avg_cost'].mean()
required_bills_flat = (GMV_TARGET_INR / 75) / average_avg_cost

print(f"\nAverage Cost for Two across targets: ₹{average_avg_cost:.2f}")
print(f"To hit exactly ₹80 Lakh with a flat distribution, you'd need to believe that on average,")
print(f"each of the 75 restaurants would process {required_bills_flat:.1f} bills per month.")

# Display target slice of prioritised prospect accounts
display_cols = ['rank', 'tier', 'name', 'avg_cost', 'is_chain', 'recommended_action', 'simulated_gmv_potential']
display(df_target_75[display_cols].head(10))
```

```python
# --- CHART 3: Restaurant Count by Tier and Cost Band (Stacked Bar) ---
# Define custom cost bands
def categorize_cost(cost):
    if cost < 400:
        return 'Below ₹400'
    elif cost <= 800:
        return '₹400-₹800'
    elif cost <= 1200:
        return '₹800-₹1200'
    else:
        return 'Above ₹1200'

# Filter to targets and categorize cost
df_target_75['cost_band'] = df_target_75['avg_cost'].apply(categorize_cost)
cost_order = ['Below ₹400', '₹400-₹800', '₹800-₹1200', 'Above ₹1200']
tier_order = ['Tier A', 'Tier B', 'Tier C']

# Create pivot count table
pivot_df = df_target_75.groupby(['cost_band', 'tier']).size().unstack(fill_value=0)
pivot_df = pivot_df.reindex(cost_order).reindex(columns=tier_order)

# Plot Stacked Bar
ax = pivot_df.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6),
    color=[palette_tiers['Tier A'], palette_tiers['Tier B'], palette_tiers['Tier C']],
    edgecolor='white',
    width=0.65
)
plt.title(f'Onboarding Targets in {TERRITORY} by Price Band and Strategic Tier', fontsize=14, fontweight='bold', pad=15, color=neutral_color)
plt.xlabel('Average Meal Cost Band (For Two)', fontsize=11, fontweight='semibold')
plt.ylabel('Number of Restaurants', fontsize=11, fontweight='semibold')
plt.xticks(rotation=0)
plt.legend(title='Account Segment Tier', frameon=True, facecolor='white')
plt.tight_layout()
plt.savefig('tier_cost_distribution.png', dpi=300)
plt.show()
plt.close()
```

### Step 11: Interpretation of Scoring Methodology and Limitations

#### Component Logic
* **Demand Score (40%)**: Measures raw dinner reviews scaled via natural log (`log1p`) to compress outliers and prevent extremely busy locations from skewing ratings.
* **Quality Score (25%)**: Scales verified dinner star ratings to prioritize premium dining experiences over sub-par culinary options.
* **Dine-In Orientation (20%)**: Computes the ratio of dinner reviews to total reviews (delivery + dinner) to prioritize locations where customers choose to visit in person.
* **Price Fit (15%)**: Uses linear interpolation to award a perfect score of 100 for average costs in our sweet spot (’400–’1200) and penalizes extremes.

#### What each component does and DOES NOT measure
* **Demand Score**: Measures past digital traction/popularity, but **does not** track current table utilization or peak hour traffic capacity.
* **Quality Score**: Reflects overall user satisfaction, but **does not** isolate subjective dining attributes like individual staff service, hygiene, or design.
* **Dine-In Orientation**: Successfully identifies locations where customers choose to visit in person, but **does not** account for spatial dining capacity or kitchen efficiency limits.
* **Price Fit**: Pinpoints partners that match the target audience's purchasing power, but **does not** reflect the gross margins or baseline food costs of the restaurant itself.
